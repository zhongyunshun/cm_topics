# 1.1 Extract topics from txt file and convert to dataframe

```python
# Initialize an empty list to hold the file name and topics data
data = []

# Directory where the txt files are located
folder_path = '/Users/yunshunzhong/Library/Mobile Documents/3L68KQB4HG~com~readdle~CommonDocuments/Documents/PhD Project/Prompt Engineering/topics_dataset/topics_results'
```

```python

# Loop through every file in the directory
for filename in os.listdir(folder_path):
    if filename.endswith('.txt'):  # Ensure the file is a text file
        file_path = os.path.join(folder_path, filename)

        # Open the text file and read its contents
        with open(file_path, 'r') as f:
            content = f.read()

            # Use a regular expression to find the topics
            match = re.search(r'[tT]opics: (.+)', content)

            if match:  # If a match was found
                topics_str = match.group(1)
                topics_list = [topic.strip() for topic in topics_str.split(',')]

                # Append the file name and topics to the data list
                data.append({'file_name': filename, 'topics': topics_list})
```

```python
# Create a DataFrame from the data
df_topics = pd.DataFrame(data)
df_topics.head()
```

If topics in the dataframe is in the value topics list of the dictionary, replace that topics in dataframe with summarized topics.

1. Create a Reverse Mapping Dictionary:
Convert topics_dict such that each original topic maps back to its summarized topic.

2. Process Each Row in the DataFrame:
For each topic in the 'topics' column of the DataFrame, check if it's in the reverse mapping dictionary. If it is, replace it with the summarized topic from the dictionary.

3. Remove Duplicates:
After processing, remove duplicate summarized topics from each list in the 'topics' column of the DataFrame.

Example dataframe input:
df = pd.DataFrame({
    'Index': [0, 1],
    'topics': [[Automated materials management, Inventory control, Construction logistics, Purchasing], [Materials management, Consolidation, Logistic, Green team risks]]
})

Example dictionary input:
topics_dict = {'Materials Management': ['Automated materials management', 'Materials management', 'Inventory control', 'Purchasing'], 'Logistics and Supply Chain Management':['Construction logistics',  'On-site logistics',  'Off-site logistics', 'Consolidation', 'Construction supply chain', 'Logistic']}

Example dataframe output:
pd.DataFrame({
    'Index': [0, 1],
    'topics': [[Materials Management, Logistics and Supply Chain Management], [Materials management, Logistics and Supply Chain Management, Green team risks]]
})

```python
# Create a reverse mapping dictionary
reverse_mapping = {}
for summarized, topics in topics_dict.items():
    for topic in topics:
        reverse_mapping[topic] = summarized

# Replace topics in dataframe with summarized topics
def replace_with_summarized(topics_in_one_paper):
    return list(set([reverse_mapping.get(topic, topic) for topic in topics_in_one_paper]))

df['topics_summazried'] = df['topics'].apply(replace_with_summarized)

df.head()
```
